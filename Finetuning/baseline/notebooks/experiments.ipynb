```python
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
import joblib
import json
import os
from datetime import datetime
from sklearn.multiclass import OneVsRestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score
```

```python
train = pd.read_csv("Finetuning/data/processed/train.csv")
dev = pd.read_csv("Finetuning/data/processed/dev.csv")
test = pd.read_csv("Finetuning/data/processed/test.csv")
```

```python
col_labels = ['toxic', 'severe_toxic', 'obscene', 'threat', 'insult', 'identity_hate']

x_train = train['comment_text'].to_list()
x_dev = dev['comment_text'].to_list()
x_test = test['comment_text'].to_list()

y_train = train[col_labels].to_numpy()
y_dev = dev[col_labels].to_numpy()
y_test = test[col_labels].to_numpy()
```

```python
vectorizer_v1 = TfidfVectorizer(
    analyzer='char_wb',
    ngram_range=(1, 3),
    min_df=3,
    max_df=0.8,
    sublinear_tf=True
)

X_train = vectorizer_v1.fit_transform(x_train)
X_dev = vectorizer_v1.transform(x_dev)
X_test = vectorizer_v1.transform(x_test)

print(f'X_train shape: {X_train.shape}')
print(f'X_dev shape: {X_dev.shape}')
print(f'X_test shape: {X_test.shape}')
```

```python
model = OneVsRestClassifier(LogisticRegression(class_weight=None), verbose=2)
model.fit(X_train, y_train)
```

```python
# How did the model do on the training set ?
# The model is struggling with the 'threat' class even with the training set.

import numpy as np
from sklearn.metrics import classification_report

# Get probabilities — OneVsRestClassifier has a single predict_proba call
y_train_proba = model.predict_proba(X_train)  # shape: (n_samples, 6)

# Custom thresholds per label — adjust these
thresholds = np.array([0.5, 0.4, 0.5, 0.4, 0.5, 0.4])

# Apply thresholds
y_train_hat = (y_train_proba >= thresholds).astype(int)

print(classification_report(y_train, y_train_hat, target_names=col_labels))
```

```python
# How did the model do on the dev set ?

# Get probabilities — OneVsRestClassifier has a single predict_proba call
y_dev_proba = model.predict_proba(X_dev)  # shape: (n_samples, 6)

# Custom thresholds per label — adjust these
thresholds = np.array([0.5, 0.4, 0.5, 0.4, 0.5, 0.4])

# Apply thresholds
y_dev_hat = (y_dev_proba >= thresholds).astype(int)

print(classification_report(y_dev, y_dev_hat, target_names=col_labels))
```

```python
# col_labels order: toxic, severe_toxic, obscene, threat, insult, identity_hate
weight_configs = [
    {"name": "no weights",        "weights": [None, None, None, None, None, None]},
    {"name": "tiny rare weights", "weights": [None, {0:1, 1:2}, None, {0:1, 1:2}, None, {0:1, 1:2}]},
    {"name": "mild rare weights", "weights": [None, {0:1, 1:3}, None, {0:1, 1:3}, None, {0:1, 1:3}]},
    {"name": "all balanced",      "weights": [{0:1, 1:2}, {0:1, 1:2}, {0:1, 1:2}, {0:1, 1:2}, {0:1, 1:2}, {0:1, 1:2}]},
]

thresholds = np.array([0.5, 0.4, 0.5, 0.4, 0.5, 0.4])

for config in weight_configs:
    models = [
        LogisticRegression(class_weight=w, max_iter=1000).fit(X_train, y_train[:, i])
        for i, w in enumerate(config["weights"])
    ]

    y_proba = np.column_stack([m.predict_proba(X_train)[:, 1] for m in models])
    y_hat = (y_proba >= thresholds).astype(int)

    print(f"\n{'='*60}")
    print(f"Config: {config['name']}")
    print('='*60)
    print(classification_report(y_train, y_hat, target_names=col_labels))
```
